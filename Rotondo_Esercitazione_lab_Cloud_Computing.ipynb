```python
%%writefile train.py
import argparse
import os
import math
from datasets import load_from_disk
from transformers import DistilBertForSequenceClassification, DistilBertTokenizerFast, Trainer, TrainingArguments
import evaluate
import numpy as np

# Risoluzione dell'eccezione MLflow impostando la variabile d'ambiente prima di importare mlflow
os.environ["MLFLOW_ALLOW_FILE_STORE"] = "true"
import mlflow

def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("--data_dir", type=str, required=True, help="Path to versioned dataset directory")
    parser.add_argument("--output_dir", type=str, required=True, help="Directory where checkpoints will be saved")
    parser.add_argument("--learning_rate", type=float, default=5e-5, help="Learning rate")
    parser.add_argument("--epochs", type=int, default=3, help="Number of training epochs")
    parser.add_argument("--batch_size", type=int, default=16, help="Batch size")
    parser.add_argument("--resume_from_checkpoint", type=str, default=None, help="Path to checkpoint to resume training from")
    args = parser.parse_args()

    # 1. Carica il dataset salvato in precedenza
    train_dataset = load_from_disk(os.path.join(args.data_dir, "train"))
    test_dataset = load_from_disk(os.path.join(args.data_dir, "test"))

    # 2. Tokenizzazione
    tokenizer = DistilBertTokenizerFast.from_pretrained("distilbert-base-uncased")
    def tokenize_function(examples):
        return tokenizer(examples["text"], padding="max_length", truncation=True)

    tokenized_train = train_dataset.map(tokenize_function, batched=True)
    tokenized_test = test_dataset.map(tokenize_function, batched=True)

    # 3. Inizializzazione Modello
    model = DistilBertForSequenceClassification.from_pretrained("distilbert-base-uncased", num_labels=2)

    # Metriche di valutazione
    metric = evaluate.load("accuracy")
    def compute_metrics(eval_pred):
        logits, labels = eval_pred
        predictions = np.argmax(logits, axis=-1)
        return metric.compute(predictions=predictions, references=labels)

    # Configurazione MLflow
    mlflow.set_tracking_uri(os.path.join(os.path.dirname(args.output_dir), "mlruns"))
    mlflow.set_experiment("sentiment-analysis-lab")

    # 4. Training Arguments (uso di eval_strategy invece di evaluation_strategy)
    training_args = TrainingArguments(
        output_dir=args.output_dir,
        learning_rate=args.learning_rate,
        num_train_epochs=args.epochs,
        per_device_train_batch_size=args.batch_size,
        per_device_eval_batch_size=args.batch_size,
        eval_strategy="epoch",
        save_strategy="epoch",
        fp16=True,
        logging_steps=10,
        load_best_model_at_end=True,
        metric_for_best_model="accuracy",
        save_total_limit=2
    )

    # 5. Inizializzazione Trainer
    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=tokenized_train,
        eval_dataset=tokenized_test,
        compute_metrics=compute_metrics,
    )

    # 6. Addestramento (con eventuale resume)
    resume = False
    if args.resume_from_checkpoint and os.path.exists(args.resume_from_checkpoint):
        resume = args.resume_from_checkpoint
        print(f"Ripristino dell'addestramento dal checkpoint: {resume}")

    trainer.train(resume_from_checkpoint=resume)

    # Salvataggio finale del modello e del tokenizer
    best_model_path = os.path.join(args.output_dir, "best_model")
    trainer.save_model(best_model_path)
    tokenizer.save_pretrained(best_model_path)
    print(f"Modello migliore salvato in: {best_model_path}")

if __name__ == "__main__":
    main()
```

```python
%%writefile serve.py
import os
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel
from transformers import pipeline

app = FastAPI(title="Sentiment Analysis Service")

# 1. Recupero configurazioni da variabili d'ambiente
MODEL_ID = os.getenv("MODEL_ID", "rotodavi/distilbert-imdb-sentiment-lab")
MODEL_REVISION = os.getenv("MODEL_REVISION", "v1")

print(f"Inizializzazione servizio. Caricamento modello: {MODEL_ID} (revisione: {MODEL_REVISION})...")
try:
    # Caricamento della pipeline di classificazione testo
    classifier = pipeline(
        "text-classification",
        model=MODEL_ID,
        revision=MODEL_REVISION
    )
    print("Modello caricato con successo!")
except Exception as e:
    print(f"Errore critico durante il caricamento del modello: {e}")
    classifier = None

class PredictionRequest(BaseModel):
    text: str

class PredictionResponse(BaseModel):
    label: str
    score: float
    model_version: str

# 2. Endpoint GET /health
@app.get("/health")
def health():
    if classifier is None:
        raise HTTPException(status_code=503, detail="Model not loaded")
    return {
        "status": "healthy",
        "model_id": MODEL_ID,
        "model_revision": MODEL_REVISION
    }

# 3. Endpoint POST /predict
@app.post("/predict", response_model=PredictionResponse)
def predict(request: PredictionRequest):
    if classifier is None:
        raise HTTPException(status_code=503, detail="Classifier pipeline not initialized")

    try:
        result = classifier(request.text)[0]
        return PredictionResponse(
            label=result["label"],
            score=float(result["score"]),
            model_version=MODEL_REVISION
        )
    except Exception as e:
        raise HTTPException(status_code=500, detail=f"Prediction error: {str(e)}")
```

```python
from transformers import pipeline

# 1. Definiamo l'ID del repository su cui è salvato il tuo modello
model_id = "rotodavi/distilbert-imdb-sentiment-lab"

# Proviamo prima con il tag v1, se fallisce usiamo il branch principale 'main'
for revision in ["v1", "main"]:
    print(f"Tentativo di caricamento del modello dal Hub: {model_id} (revisione: {revision})...")
    try:
        classifier = pipeline("text-classification", model=model_id, revision=revision)
        print(f"Modello caricato con successo usando la revisione '{revision}'!\n")
        break
    except Exception as e:
        print(f"Revisione '{revision}' non disponibile o errore: {e}")
else:
    classifier = None
    print("Impossibile caricare il modello da nessuna delle revisioni provate.")

if classifier is not None:
    # 2. Test con recensioni personalizzate
    positive_review = "This movie was a masterpiece! Outstanding acting and brilliant direction."
    negative_review = "What a waste of time. The plot made no sense and the acting was terrible."

    res_pos = classifier(positive_review)[0]
    res_neg = classifier(negative_review)[0]

    # 3. Visualizzazione dei risultati
    print("\n--- RISULTATI DEL TEST ---")
    print(f"Test Recensione Positiva:")
    print(f"  Testo: \"{positive_review}\"")
    print(f"  Risultato: Label={res_pos['label']}, Score={res_pos['score']:.4f}\n")

    print(f"Test Recensione Negativa:")
    print(f"  Testo: \"{negative_review}\"")
    print(f"  Risultato: Label={res_neg['label']}, Score={res_neg['score']:.4f}")
```
